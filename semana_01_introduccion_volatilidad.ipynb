```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Serie de precios
precios = [100, 101, 100.5, 102, 103, 101, 104, 105]


# Crear una tabla con los precios
df = pd.DataFrame({
    "Precio": precios
})
```

```python
#Rendimiento logarítmico

df["Rendimiento"] = np.log(
    df["Precio"] / df["Precio"].shift(1)
)


print("\nTabla de precios y rendimientos:\n")
print(df)

#Gráfica de precios

plt.figure(figsize=(8, 4))

plt.plot(
    df["Precio"],
    marker="o"
)

plt.xlabel("Día")
plt.ylabel("Precio")
plt.title("Serie de precios")

plt.grid()

plt.show()

#Gráfica de rendimientos logarítmicos

plt.figure(figsize=(8, 4))

plt.plot(
    df["Rendimiento"],
    marker="o"
)

plt.axhline(0)

plt.xlabel("Día")
plt.ylabel("Rendimiento")
plt.title("Rendimientos logarítmicos")

plt.grid()

plt.show()
```

```python
#ACTIVIDAD 1

print("ACTIVIDAD 1")

precios_1 = [100, 102, 101, 105, 103]


actividad1 = pd.DataFrame({
    "Precio": precios_1
})

#Cambio absoluto

actividad1["Cambio absoluto"] = (
    actividad1["Precio"]
    - actividad1["Precio"].shift(1)
)

#Rendimiento simple

actividad1["Rendimiento simple"] = (
    (
        actividad1["Precio"]
        - actividad1["Precio"].shift(1)
    )
    /
    actividad1["Precio"].shift(1)
)

#Rendimiento logarítmico

actividad1["Rendimiento logarítmico"] = np.log(
    actividad1["Precio"]
    /
    actividad1["Precio"].shift(1)
)

print("\nTabla completa:\n")

print(actividad1)

# Rendimientos positivos

positivos = actividad1[
    actividad1["Rendimiento logarítmico"] > 0
]


print("\nDías con rendimiento positivo:\n")

print(
    positivos[
        [
            "Precio",
            "Rendimiento logarítmico"
        ]
    ]
)

# Rendimientos negativos

negativos = actividad1[
    actividad1["Rendimiento logarítmico"] < 0
]


print("\nDías con rendimiento negativo:\n")

print(
    negativos[
        [
            "Precio",
            "Rendimiento logarítmico"
        ]
    ]
)

# Comparación entre rendimiento simple
# y rendimiento logarítmico

print("\nComparación entre rendimientos:\n")

print(
    actividad1[
        [
            "Rendimiento simple",
            "Rendimiento logarítmico"
        ]
    ]
)

# Gráfica de comparación

plt.figure(figsize=(8, 4))


plt.plot(
    actividad1["Rendimiento simple"],
    marker="o",
    label="Rendimiento simple"
)


plt.plot(
    actividad1["Rendimiento logarítmico"],
    marker="o",
    label="Rendimiento logarítmico"
)


plt.axhline(0)

plt.xlabel("Día")
plt.ylabel("Rendimiento")
plt.title("Rendimiento simple vs logarítmico")

plt.legend()

plt.grid()

plt.show()
```

```python
#ACTIVIDAD 2

print("ACTIVIDAD 2")

A = np.array([
    0.1,
    -0.2,
    0.2,
    -0.1,
    0.1,
    -0.1
])


B = np.array([
    3,
    -4,
    5,
    -2,
    4,
    -5
])


print("\nSerie A:")
print(A)


print("\nSerie B:")
print(B)

#Desviación estándar

desviacion_A = np.std(
    A,
    ddof=1
)


desviacion_B = np.std(
    B,
    ddof=1
)


print("\nDesviación estándar de A:")
print(desviacion_A)


print("\nDesviación estándar de B:")
print(desviacion_B)

# Determinar cuál tiene mayor volatilidad

if desviacion_A > desviacion_B:

    print(
        "\nLa serie A tiene mayor volatilidad."
    )

else:

    print(
        "\nLa serie B tiene mayor volatilidad."
    )

# Gráfica de ambas series

plt.figure(figsize=(8, 4))


plt.plot(
    A,
    marker="o",
    label="Serie A"
)


plt.plot(
    B,
    marker="o",
    label="Serie B"
)


plt.axhline(0)

plt.xlabel("Observación")
plt.ylabel("Rendimiento")

plt.title(
    "Comparación visual de volatilidad"
)

plt.legend()

plt.grid()

plt.show()
```

```python
#ACTIVIDAD 3

print("ACTIVIDAD 3")
print("========================================")


r = np.array([
    0.2,
    -0.1,
    0.3,
    -0.2,
    0.1,
    -0.3,
    3.2,
    -4.1,
    2.8,
    -3.5,
    4.0,
    -2.9,
    0.4,
    -0.2,
    0.1,
    -0.3
])


print("\nSerie de rendimientos:\n")

print(r)

#Crear número de observaciones

observaciones = np.arange(
    1,
    len(r) + 1
)

# Gráfica de la serie

plt.figure(figsize=(10, 4))


plt.plot(
    observaciones,
    r,
    marker="o"
)


plt.axhline(0)


plt.xlabel("Observación")

plt.ylabel("Rendimiento")

plt.title(
    "Regímenes de volatilidad"
)


plt.grid()

plt.show()

#Identificación visual de los regímenes

regimenes = [

    "Baja",
    "Baja",
    "Baja",
    "Baja",
    "Baja",
    "Baja",

    "Alta",
    "Alta",
    "Alta",
    "Alta",
    "Alta",
    "Alta",

    "Baja",
    "Baja",
    "Baja",
    "Baja"
]



#Crear tabla
actividad3 = pd.DataFrame({

    "Observación": observaciones,

    "Rendimiento": r,

    "Régimen": regimenes

})


print("\nClasificación visual:\n")

print(actividad3)

# Respuestas de la actividad

print(
    "\n1. El periodo de alta volatilidad "
    "parece comenzar en la observación 7."
)


print(
    "\n2. El periodo de alta volatilidad "
    "parece terminar en la observación 12."
)


print(
    "\n3. La característica utilizada para "
    "identificarlo fue el tamaño de los "
    "rendimientos."
)


print(
    "\n4. Con 3000 observaciones sería muy "
    "difícil realizar esta clasificación "
    "manualmente, por lo que sería útil "
    "emplear métodos automáticos."
)
```
